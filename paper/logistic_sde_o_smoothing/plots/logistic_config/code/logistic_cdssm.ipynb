```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import plots

data = plots.load('logistic_config.json')
```

#  Plot of the Logistic CD-SSM and its Lamperti Transform

```python
fig, axes = plots.logistic_cdssm()

fig.tight_layout()
fig.savefig(
    "../files/logistic_cdssm.png",
    dpi=300,
    bbox_inches="tight"
)
```

# CPU Times of the Algorithms

```python
fig, ax = plt.subplots(figsize=(7.2, 4.8))
ax = plots.cpu_times_boxplot(ax=ax, data=data)

fig.tight_layout()
fig.savefig(
    "../files/cpu_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
