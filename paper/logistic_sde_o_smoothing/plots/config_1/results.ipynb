# 1. The Model: One Dimensional OU-CDSSM

We assume the OU CD-SSM, which is consists of a continuous-time latent Ornstein-Uhlenback process:

\begin{equation}
dX(s) = -\rho X(s) ds + \phi X(s) \qquad X(0) = 0
\end{equation}

We have $T$ observations at the discrete times $s_t = t: t \in \mathcal{T}$, with $\Delta_t=1$ with noise:

\begin{equation}
Y_t = X(s_t) + \epsilon_t 
\end{equation}

Where $\epsilon_t \sim \mathcal{N}(0, \sigma_Y^2)$ is an independent innovations process. We fix the parameter vector $\theta = (\rho, \phi, \sigma_Y) = (1, 1, 0.2)$
