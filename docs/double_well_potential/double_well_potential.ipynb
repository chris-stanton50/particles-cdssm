```python
from math import sqrt
import numpy as np
import matplotlib.pyplot as plt
from functools import partial
```

# The Double Well Potential

In this notebook, we consider a classical example of an SDE that is motivated by intuition that is drawn from Physics: we consider the double-well potential.
We begin with its ODE form: the potential function $U: \mathbb{R} \rightarrow \mathbb{R}$ is given by:
%
\begin{equation}
U(x) = a x^4 - b x^2
\end{equation}
%
This function has the following properties:

- Even function
- 3 Stationary Points
- 3 Roots (incl. one repeated root)

We plot the double-well potential below:

```python
def U(x, a=1., b=1.):
    """
    The double well potential:
    """
    return a * np.power(x, 4) - b * np.power(x, 2)    
```

```python
a = 1.; b=1.
start=-1.2*np.sqrt(b/a); stop=1.2*np.sqrt(b/a)

x = np.linspace(start=start, stop=stop, num=1000)
potential = U(x)

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)

dwp_line2d_kwargs = dict(c='blue', lw=1., ls='-', marker=None, alpha=1.0)
line = ax.plot(x, potential, **dwp_line2d_kwargs)

ax.set(xlabel=r'$x$', ylabel=r'$U(x)$', title='Double Well Potential', xlim=(start, stop), ylim=(np.min(potential), np.max(potential)))
ax.set_xticks([-sqrt(b/a), -sqrt(b/(2*a)), 0., sqrt(b/(2*a)), sqrt((b/a))])
ax.set_xticklabels([r'$-\sqrt{\frac{b}{a}}$', r'$-\sqrt{\frac{b}{2a}}$', 0., r'$\sqrt{\frac{b}{2a}}$', r'$\sqrt{\frac{b}{a}}$'])


x_axis_line2d_kwargs = dict(lw=2.0, ls='-', color='black', alpha=1.0)
ax.vlines()
ax.hlines(0., xmin=start, xmax=stop, **x_axis_line2d_kwargs)
plt.show()
```
